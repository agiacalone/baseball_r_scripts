```python
import sys
sys.path.insert(0, "..")
from src.baseball import *
```

```python
# Parameters — change TEAM_ID and TEAM_NAME to use a different team
SEASON     = 2025
TEAM_ID    = 137        # San Francisco Giants
TEAM_NAME  = "Giants"
OUTPUT_DIR = f"../output/{TEAM_NAME}"

today     = date.today()
yesterday = today - timedelta(days=1)
tomorrow  = today + timedelta(days=1)
```

```python
# Resolve game IDs
today_game_id     = game_ident(TEAM_ID, today)
yesterday_game_id = game_ident(TEAM_ID, yesterday)
tomorrow_game_id  = game_ident(TEAM_ID, tomorrow)

print(f"Today's game:     {today_game_id}")
print(f"Yesterday's game: {yesterday_game_id}")
print(f"Tomorrow's game:  {tomorrow_game_id}")
```
